### Q3. Check for Outliers by finding distributions of the data both in Categorical and Numerical Columns.

```python
# First we Find the distributions of data in either Categorical column Versus Numerical Columns
## Numerical_Columns
for col in numerical_cols: 
    plt.figure(figsize=(10,8))
    sns.histplot(job_df[col], kde=True, bins=30)
    plt.title(f"Distribution of {col}")
    plt.xlabel(col)
    plt.ylabel("Frequency")
    plt.show()
```

```python
## Check if there any outliers in Numerical Columns:
for col in numerical_cols:
    plt.figure(figsize=(8,4))
    sns.boxplot(x=job_df[col])
    plt.title(f"Boxplot of {col}")
    plt.show()
```

```python
## Check for distributions in Categorical Columns
for col in categorical_cols:
    top_n = job_df[col].value_counts().nlargest(10)
    
    plt.figure(figsize=(10,8))
    sns.barplot(x=top_n.index, y=top_n.values)
    plt.title(f"Top categories in {col}")
    plt.xticks(rotation=45)
    plt.show()
```

### Q4. What is the avearge salary for the tech job roles?

```python
Average_salary = job_df['salary'].mean().round(0)
Average_salary
```

### Q5. How does the average salary compare to the level of education?

```python
salary_education_comparison = job_df.groupby('education_level').agg(
    avg_salary = ('salary','mean'))
salary_education_comparison

salary_education_comparison.sort_values('avg_salary',ascending=False).round(0)

## Result; Employees with the Highest Eduction level(PhD) earned the most compared to those with lower levels of education
```

### Q6. How does the Average Salary compare to the Level of Experience

```python
salary_explevel_comparison = job_df.groupby('experience_years').agg(
    avg_salary =('salary','mean')).sort_values('avg_salary',ascending = False).round(0)
salary_explevel_comparison

## Result: Employees with the highest levels of Expereince(over 10years) earned the most amount of money compared to those with lower experience in years,
```

### Q7. Is there any correlation between experience in years and the salary?

```python
exper_salary_corr = job_df['experience_years'].corr(job_df['salary']).round(4)
print(exper_salary_corr)
    ##Result: Implies there is a moderate relationship in having years of experience and a higher salary.
```

### Q8. Is there any correlation between salary and the number of certifications?

### Q9. Which industry and job title earn the highest salary?


```python
most_salary = job_df.groupby(['job_title', 'industry']).agg(
    avg_salary =('salary','mean')).sort_values('avg_salary',ascending = False).round(0)
most_salary

## Result: AI Engineers working in Media, Healthcare and Finance are the Highest earners with a salary of 174K
```

### Q10. Which job title and industries earn thee least?

```python
most_salary.tail(10)

## Data Analysts working in Finance, Consulting, Healthcare and Manufacturing are the least earners at an average salary of 120K each.
```

### Q11. According to the company category, which dominates in size?

```python
company_breakdown = job_df.groupby('company_size')['company_size'].value_counts()
company_breakdown

company_breakdown.sort_values(ascending=False)

## Result: Employees are equally distributed across all the 4 categories in regards to the company size. No single category is SIGNIFICANTLY DOMINANT over the rest.
## However for the company sized "Large" has the highest number of people employed at 50254.
```

### Q12.How does having a certification affect salary whean controlling for experience in years?

```python
certs_salary = job_df.groupby(['experience_years', 'certifications']).agg(
    avg_salary =('salary','mean'))
certs_salary
plt.figure(figsize=(10,6))
sns.lineplot(
    data= certs_salary,
    x='experience_years',
    y='avg_salary',
    hue='certifications'
)

plt.title('Average Salary Growth by Experience Controlling for Certifications')
plt.show()


## Result: Employees who had numerous certifications earned a hihgher salary compared to those with Zero on One certifications.
```

### Q13. How are various job titles and industries spread across the locations?

```python
## Q8. How are the various Job titles and Industries spread across the locations
## Create a frequency table to view our data for Job Title Versus Location
job_loc = pd.crosstab(job_df['location'], job_df['job_title'])
print(job_loc)
```

```python
## Visualize how the job titles spread across the locations
plt.figure(figsize=(12,8))
sns.heatmap(job_loc, cmap='coolwarm')

plt.title('Job Title Distribution Across Locations')
plt.xlabel('job_title')
plt.ylabel('location')
plt.show()
```

### Q14. Distribution of Industry  versus Location

```python
indu_loc = pd.crosstab(job_df['location'], job_df['industry'])
print(indu_loc)
```

```python
# To visualize our Industry vs location using a heat map
plt.figure(figsize=(12,8))
sns.heatmap(indu_loc, cmap='viridis')

plt.title('Industry Distribution Across Locations')
plt.xlabel('industry')
plt.ylabel('location')
plt.show()
```
